```python
# savefig("filename.extension", dpi= value, bbox_inches='tight')

# filename.extension: The name of the file to save the figure to, including the desired file extension (e.g., .png, .jpg, .pdf).
# dpi: Dots per inch, which controls the resolution of the saved figure. Higher values result in better quality but larger file sizes.
# bbox_inches: This parameter can be set to 'tight' to ensure that the bounding box of the figure is tight around the content, minimizing extra whitespace.
import matplotlib.pyplot as plt

x=[1, 2, 3, 4, 5]
y=[20, 30, 50, 70, 110]

# creating the plot 
plt.plot(x ,y , color='red', linestyle='dashed', linewidth = 3,
         marker='o', markerfacecolor='blue', markersize=12)
plt.title('Sample graph')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

plt.savefig("samplegraph.png", dpi=300, bbox_inches='tight')
# savefig() function in Matplotlib is used to save the current figure to a file.
plt.show()


# note --> The savefig() function should be called before plt.show() to ensure that the figure is saved correctly.

```
