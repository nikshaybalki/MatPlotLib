```python
import matplotlib.pyplot as plt
```

  ## scatter plot 

```python
# plt.scatter( x,y, color='color name' , marker = 'marker style' , label = 'label name')

hour= [2,4,6,8,10,12]

score = [40,60,70,50,44,80]

plt.scatter(hour, score , color='blue' , marker='o' , label="student data")

plt.xlabel("hours")
plt.ylabel("scores")
plt.title("study hour and score relationship")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# plt.scatter( x,y, color='color name' , marker = 'marker style' , label = 'label name')


plt.scatter([2,4,6], [50,52,55] , color='blue' , marker='o' , label="class A")
plt.scatter([2,4,6], [45,50,53] , color='green' , marker='o' , label="class B")


plt.xlabel("hours")
plt.ylabel("scores")
plt.title("study hour and score relationship between 2 classes")
plt.legend()
plt.grid(True)
plt.show()
```
