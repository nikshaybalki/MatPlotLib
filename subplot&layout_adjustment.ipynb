```python
import matplotlib.pyplot as plt
```

```python
# plt.subplot(nrows , ncols , index) ## in this indexing start from 1 .

x=[2,3,4,6]

y=[33,54,34 , 43]

plt.subplot(1,2,1)  # 1 row , 2 column , 1 st subplot
plt.plot(x,y)
plt.title("graph-1")

plt.subplot(1,2,2)  # 1 row , 2 column , 2 st subplot
plt.bar(x,y)
plt.title("graph-2")


plt.tight_layout()
plt.show()
```

```python
# fig ,ax = plt.subplots(nrows , ncols ,figsize=(width,height))
fig ,ax = plt.subplots(1 , 2 ,figsize=(10,5))


x=[2,3,4,6]

y=[33,54,34 , 43]

ax[0].plot(x,y , color="red")
ax[0].set_title('Line plot')

ax[1].bar(x,y ,color="green")
ax[1].set_title('Bar chart')


fig.suptitle("My Graphs" )


plt.tight_layout()

plt.show()
```
