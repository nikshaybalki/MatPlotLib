```python
import matplotlib.pyplot as plt
```

```python
 # plt.plot(x,y, color='color name' , linestyle='line_style' ,
 #  linewidth=value(in integer) , marker='marker symbol' , label='label name')

month=[1,2,3,4] 

sales=[100,130,24,133]
 
plt.plot(month ,sales , color='orange' , linestyle='--' , linewidth=2 , marker='o' ,label='sales'  )
plt.xlabel('month')
plt.ylabel('sales')
plt.title("report")

plt.legend( loc='lower right' , fontsize=12)

plt.grid(color='gray' , linestyle=':' , linewidth=1 )

plt.xlim(1,4)  # x-limit kitne interval tak hogi 
plt.ylim(0,150)   # y-limit kitne interval tak hogi 

plt.xticks([1,2,3,4] , ['mon' , 'tue' , 'wed' , 'thu'])

plt.show(1,4) 


```

## plt.bar()

```python
#plt.bar(x , height , color='color name' , width = value , label = 'label name' )

product = ["chatgpt" , "gemini" , "grok" , "perplexity" ]

download =[100 ,80 , 60 ,120]

# plt.barh(product , download , color="purple" , label='download 2025') --> show the data horizontally
plt.barh(product , download , color="purple" , label='download 2025')

plt.xlabel("APP")
plt.ylabel("monthly download")
plt.title("download report")
plt.legend()
plt.show()
```

## plt.pie

```python
#plt.pie(values ,labels=label_list , colors= color_list , autopct = '%1.1f%%' )

regions=["North" , "South" , "East" , "West"]
reveneu = [2000 , 2500 , 1000 , 5000]

#  autopct = '%1.1f%%' --> regionwise chart me % show karne ke liye

plt.pie(reveneu , labels=regions ,autopct = '%1.1f%%',   colors=['green' , 'orange' , 'yellow' , 'blue' ])
plt.title("region wise contribution")
plt.show()
```
